```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model = "gemini-3.6-flash")
prompts = [
    {"role":"system","content":"You are JavaScript Developer"},
    {"role":"user","content":"Difference b/w map and filter?"}
]

res = llm.invoke(prompts)
print(res.content[0]["text"])
```

#### Prompts Templates - Dynamic Prompts

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")
out = StrOutputParser()
```

```python
prompts = ChatPromptTemplate.from_messages( [
    {"role":"system","content":"You are a translator and translate input in {language}"},
    {"role":"user","content":"{query}"}
])
```

```python
## Myself chain node

def transform_case(result:str):
    return result.upper()
```

```python
chains = prompts | llm | out | transform_case
```
