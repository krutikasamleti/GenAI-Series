#### Static Prompts

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash"
)

prompts = [
    ("system", "You are a Java Developer."),
    ("user", "How to sort an array?")
]

res = llm.invoke(prompts)

print(res.content)
```
